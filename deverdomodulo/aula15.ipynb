```python
from math import sqrt

while True:
    N = int(input("Por favor, digite um número: "))
    if N < 0:
        print("Número inválido")
        continue
    elif N == 0:
        break
    print(sqrt(N))
print("Saindo...")
```

```python
senha = input("Senha: ")

while True:
    senha2 = input("Repita a senha: ")
    if senha2 == senha:
        print("Conta de usuário criada!")
        break
    elif senha2 != senha:
        print("As senhas não coincidem!")
```

```python
pin1 = int(4321)
t = 1

while True:
    pin2 = int(input("PIN: "))
    if pin2 == pin1:
        print(f"Correto! Você precisou de {t} tentativas")
        break
    elif pin2 != pin1:
        print("Errado")
        t += 1
```

```python
# ----------------- próximo ano bissexto ---------------------

ano = int(input("Ano: "))
ano1 = ano

while True:
    ano1 += 1
    if (ano1 % 4 == 0 and ano1 % 100 != 0) or (ano1 % 400 == 0):
        print(f"O próximo ano bissexto após {ano} é {ano1}")
        break
```

```python
hist = ""
palavra2 = ""

while True:
    palavra = input("Por favor, digite uma palavra: ")
    if palavra == "fim":
        print(hist)
        break
    elif palavra == palavra2:
        print(hist)
        break
    hist += palavra + " "
    palavra2 = palavra
```

```python


print("Por favor, digite números inteiros. Digite 0 para terminar.")
N_P = 0
N_N = 0
soma = 0
media = 0
i = 0
while True:
    numero = int(input("Número: "))
    soma += numero
    if numero == 0:
        media = soma / (i)  # ou media = soma / i if i > 0 else 0
        print(f"Números digitados {i}")
        print(f"A soma dos números é {soma}")
        print(f"A média dos números é {media}")
        print(f"Números positivos {N_P}")
        print(f"Números negativos {N_N}")
        break
    elif numero > 0:
        N_P += 1
    else:
        N_N += 1
    i += 1
```
