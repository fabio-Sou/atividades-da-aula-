```python
# 10)-------------------- __ --------------------

my_list = [1, 3, 4]
N = input("Digite um número: ")

if int(N) in my_list:
    print("A lista contém o item " + N)

if 2 in my_list:
    print("A lista contém o item 2")
else:
    print("A lista não contém o item 2")
```

```python
# 11)--------------- Ordenando listas ---------------
# sort muda a ordem da lista original
# enquanto sorted cria uma nova cópia ordenada da lista

my_list = [2,5,1,2,4]
my_list.sort()
print(my_list)
```
