```python
N = int(input("Limite: "))
SomNum = 1
SomNum1 = 0

while SomNum1 < N:
    SomNum1 += SomNum
    SomNum += 1

print(SomNum1)
```

```python
N = float(input("Limite: "))
SomNum = 1
SomNum1 = 0 
parcelas = []

while SomNum1 < N:
    parcelas.append(str(SomNum))
    SomNum1 += SomNum
    SomNum += 1

print(f"A soma consecutiva: {' + '.join(parcelas)} = {SomNum1}")
```
