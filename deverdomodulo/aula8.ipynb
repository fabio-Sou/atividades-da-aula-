```python
km = float(input("Distância (km): "))

if km <= 0:
 print("Distância inválida")
elif km <= 50:
 total = km * 1.75
elif km <= 150:
 total = 50 * 1.75 + (km - 50) * 1.65
else:
 total = 50 * 1.75 + 100 * 1.65 + (km - 150) * 1.50

if km > 0:
 medio = total / km
 print(f"Total: R$ {total:.2f}")
 print(f"Médio: R$ {medio:.2f}/km")
```
