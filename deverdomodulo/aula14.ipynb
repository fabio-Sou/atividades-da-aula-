```python
# ---------------- Calculo de Parte Inteira e Decimal 1.2 -----------------

s = input("Por favor, digite uma temperatura: ")
s = s.replace(',', '.')        # aceita vírgula na entrada
x = float(s)
int_part = int(x)
dec_part = abs(x - int_part)

dec_str = f"{dec_part:.2f}".rstrip('0').rstrip('.')   # remove zeros à direita
if '.' not in dec_str:
    dec_str += '.0'                                   # garante ao menos 1 casa
dec_str = dec_str.replace('.', ',')                  # usa vírgula como separador
if dec_str.startswith(','):
    dec_str = '0' + dec_str                          # garante o "0" antes da vírgula

print(f"Parte inteira: {int_part}")
print("Parte decimal: " + dec_str)
```

```python
# ----------------- ordem alfabética das palavras ------------------

p1 = input("Por favor, digite a 1ª palavra: ").strip()
p2 = input("Por favor, digite a 2ª palavra: ").strip() # .strip() faz com que espaços em branco no início e no final da palavra sejam removidos.

if p1 == p2:
    print("Você digitou a mesma palavra duas vezes.")
elif p1 > p2:
    print(f"{p1} vem por último em ordem alfabética.")
else:
    print(f"{p2} vem por último em ordem alfabética.")
```

```python
# ----------------- Qual é a letra do meio 1.1 -----------------

letras = []
for i in range(1, 4): # faz 3 repetições (1, 2, 3)  # * Começa em 1 * Termina ANTES de 4 (ou seja, para no 3) * Resultado: 1, 2, 3
    letras.append(input(f"{i}ª letra: ")) # coleta cada letra e adiciona à lista

meio = sorted(letras)[1] # ordena as letras e pega a do meio (índice 1)
print("A letra do meio é", meio)
```

```python
# --------- Calculadora de Imposto sobre Doação ---------

Val_Da_Doaç = float(input("Valor da doação: "))

# Tabela de faixas de imposto
# [lim_inf_doaç, lim_sup_doaç, imposto_base, alíquota]
Faixas = [
    (5000, 25000, 100, 8),
    (25000, 55000, 1700, 10),
    (55000, 200000, 4700, 12),
    (200000, 1000000, 22100, 15),
    (1000000, float('inf'), 142100, 17) # float('inf') representa o infinito em Python
]

imposto = 0
encontrado = False

for Lim_Inf, Lim_Sup, Impo_B, Aliquota in Faixas:
    if Lim_Inf <= Val_Da_Doaç < Lim_Sup:
        imposto = Impo_B + (Val_Da_Doaç - Lim_Inf) * Aliquota / 100
        encontrado = True
        break
    elif Val_Da_Doaç >= Lim_Sup and Lim_Sup == 1000000:
        imposto = Impo_B + (Val_Da_Doaç - Lim_Inf) * Aliquota / 100
        encontrado = True
        break

if encontrado:
    print(f"Valor do imposto: {imposto:.1f} euros")
else:
    print("Sem imposto!")
```
