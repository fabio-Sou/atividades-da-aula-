# Calculadora em Python

## Uma melhoria da calculadora original

**Objetivo:** fazer as quatro operações básicas, validar entradas e permitir vários cálculos sem reiniciar o programa.

**Tempo de apresentação:** cerca de 12 a 15 minutos.

O projeto usa apenas assuntos já trabalhados nas atividades: `input`, `print`, variáveis, conversão para `float`, condicionais, laços `while`, funções, `try/except` e f-strings. Não usa bibliotecas adicionais.

## Roteiro para apresentar

1. **1 min — Problema:** a versão original fazia uma conta e encerrava.
2. **3 min — Função `calcular`:** separa a operação e decide com `if/elif/else`.
3. **3 min — Função `ler_numero`:** converte o texto digitado e repete a pergunta se não for número.
4. **4 min — Função `main`:** organiza as perguntas e repete os cálculos.
5. **3 min — Demonstração:** conta normal, número inválido, divisão por zero e saída.

A apresentação pode ser encurtada pulando os testes de erro ou mostrando só um deles.

## 1. Separando o cálculo da conversa

A função recebe dois números e o símbolo da operação. `return` devolve a resposta. A divisão por zero e os símbolos desconhecidos geram um erro com uma mensagem clara.

```python
def calcular(valor1, valor2, operacao):
    if operacao == "+":
        return valor1 + valor2
    elif operacao == "-":
        return valor1 - valor2
    elif operacao == "*":
        return valor1 * valor2
    elif operacao == "/":
        if valor2 == 0:
            raise ValueError("Não é possível dividir por zero.")
        return valor1 / valor2
    else:
        raise ValueError("Operação inválida. Use +, -, * ou /.")


print("10 + 5 =", calcular(10, 5, "+"))
print(f"10 / 5 = {calcular(10, 5, '/'):g}")
```

## 2. Lendo números com segurança

`input` sempre recebe texto. Por isso, removemos espaços, trocamos vírgula decimal por ponto e tentamos converter com `float`. Se a conversão falhar, `except` mostra a orientação e `while` pergunta novamente.

```python
def ler_numero(mensagem):
    while True:
        entrada = input(mensagem).strip().replace(",", ".")
        try:
            return float(entrada)
        except ValueError:
            print("Digite um número válido. Exemplo: 3 ou 2,5.")
```

## 3. Juntando as partes e repetindo a calculadora

`main` conduz a conversa. O laço continua até a pessoa digitar `q` ou responder que não quer outra conta. `try/except` captura os erros previstos na função de cálculo sem encerrar a calculadora.

**Demonstração sugerida:** faça `10 / 4`, depois digite `abc` como número para mostrar a validação; teste `8 / 0`; por fim, digite `q` para sair. Execute esta célula e responda às perguntas no terminal interativo do notebook.

```python
def main():
    print("=== Calculadora ===")
    print("Operações disponíveis: +, -, * e /")
    print("Digite q no campo da operação para sair.\n")

    while True:
        valor1 = ler_numero("Primeiro número: ")
        operacao = input("Operação (+, -, *, /) ou q para sair: ").strip().lower()

        if operacao == "q":
            print("Calculadora encerrada.")
            break

        valor2 = ler_numero("Segundo número: ")

        try:
            resultado = calcular(valor1, valor2, operacao)
        except ValueError as erro:
            print("Erro:", erro)
        else:
            print(f"Resultado: {resultado:g}")

        continuar = input("Deseja fazer outra conta? (s/n): ").strip().lower()
        if continuar != "s":
            print("Calculadora encerrada.")
            break


main()
```
