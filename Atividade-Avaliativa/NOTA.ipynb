```python
# NOTA DA QUESTÃO 3:

# Solicita um número ao usuário como texto
s = input("Por favor, digite um número: ")

# Converte o texto para número decimal
x = float(s)

# Obtém a parte inteira do número
int_part = int(x)

# Calcula a parte decimal usando valor absoluto
dec_part = abs(x - int_part)

# Formata a parte decimal com até 4 casas
# Depois, remove zeros e pontos desnecessários
dec_str = f"{dec_part:.4f}".rstrip("0").rstrip(".")

# Se não houver casas decimais, adiciona ".0"
if "." not in dec_str:
    dec_str += ".0"

# Garante que números menores que 1 apareçam como "0.x"
if dec_str.startswith("."):
    dec_str = "0" + dec_str

# Exibe a parte inteira e a parte decimal
print(f"Parte inteira: {int_part}")
print("Parte decimal: " + dec_str)
```

```python
# no if da para se usar "and" para colocar mais de uma condição, por exemplo:

I = 6
if I > 5 and I < 100:
    print("O número está entre 5 e 100")
# também é possível usar "or" para colocar mais de uma condição, por exemplo:

if I < 5 or I > 100:
    print("O número está fora do intervalo de 5 a 100")
```

```python
# também é possível usar "not" para negar uma condição, por exemplo:

I = 6
if not I > 5:
    print("O número não é maior que 5")
```

```python
# também é possível usar "in" para verificar se um valor está em uma lista, por exemplo:

I = 6
if I in [1, 2, 3, 4, 5]:
    print("O número está na lista")
    
# também é possível usar "not in" para verificar se um valor não está em uma lista, por exemplo:

if I not in [1, 2, 3, 4, 5]:
    print("O número não está na lista")
```

```python
# também é possível usar "is" para verificar se um valor é igual a outro, por exemplo:

I = 6
# "is" verifica se duas variáveis apontam para o mesmo objeto.
# Para comparar valores, use "==".
# Exemplo recomendado:
if I == 5:
    print("O número é igual a 5")

# "is not" verifica se duas variáveis não apontam para o mesmo objeto.
# Para verificar valores diferentes, use "!=".
if I != 5:
    print("O número não é igual a 5")

# "is" é usado principalmente para verificar None:
valor = None

if valor is None:
    print("Nenhum valor foi informado")

if valor is not None:
    print("Um valor foi informado")
```

```python
# "break" "continue" e "pass" são palavras reservadas do Python que controlam o fluxo de execução de loops. Você pode usá-las assim:
# break, continue e pass controlam a execução de um laço.

for numero in range(1, 6):
    # pass não faz nada; serve como espaço reservado
    if numero == 1:
        pass

    # continue pula para a próxima repetição
    if numero == 3:
        continue

    # break encerra o laço completamente
    if numero == 5:
        break

    print(numero)
```
