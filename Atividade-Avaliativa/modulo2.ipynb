```python
# Questão 3

s = input("Por favor, digite um número: ")
x = float(s)
int_part = int(x)
dec_part = abs(x - int_part)

dec_str = f"{dec_part:.4f}".rstrip('0').rstrip('.')
if '.' not in dec_str:
    dec_str += '.0'
if dec_str.startswith(','):
    dec_str = '0' + dec_str

print(f"Parte inteira: {int_part}")
print("Parte decimal: " + dec_str)
```

```python
# Questão 14

Val_Da_Doaç = float(input("Valor da doação: "))

Faixas = [
    (5000, 25000, 100, 8),
    (25000, 55000, 1700, 10),
    (55000, 200000, 4700, 12),
    (200000, 1000000, 22100, 15),
    (1000000, float('inf'), 142100, 17)
]

imposto = 0
encontrado = False

for Lim_Inf, Lim_Sup, Impo_B, Aliquota in Faixas:
    if Lim_Inf <= Val_Da_Doaç < Lim_Sup:
        imposto = Impo_B + (Val_Da_Doaç - Lim_Inf) * Aliquota / 100
        encontrado = True
        break
    elif Val_Da_Doaç >= Lim_Sup and Lim_Sup == 1000000:
        imposto = Impo_B + (Val_Da_Doaç - Lim_Inf) * Aliquota / 100
        encontrado = True
        break

if encontrado:
    print(f"Valor do imposto: {imposto:.1f} euros")
else:
    print("Sem imposto!")
```

```python
# Questão 19

pin1 = int(4321)
t = 1

while True:
    pin2 = int(input("PIN: "))
    if pin2 == pin1:
        if t == 1:
            print("Correto! Você precisou de apenas uma única tentativa!")
            break
        else:
            print(f"Correto! Você precisou de {t} tentativas")
            break
    elif pin2 != pin1:
        print("Errado")
        t += 1
```

```python
# Questão 22

print("Por favor, digite números inteiros. Digite 0 para terminar.")
N_P = 0
N_N = 0
soma = 0
media = 0
i = 0
while True:
    numero = int(input("Número: "))
    soma += numero
    i += 1
    if numero == 0:
        media = soma / (i - 1)  # ou media = soma / i if i > 0 else 0
        print(f"Números digitados {i - 1}")
        print(f"A soma dos números é {soma}")
        print(f"A média dos números é {media}")
        print(f"Números positivos {N_P}")
        print(f"Números negativos {N_N}")
        break
    elif numero > 0:
        N_P += 1
    else:
        N_N += 1
```
