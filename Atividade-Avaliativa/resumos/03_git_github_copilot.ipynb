## 1. Para que servem Git e GitHub?

- **Git** registra versões do projeto localmente e permite comparar mudanças.
- **GitHub** hospeda repositórios Git para cópia remota, colaboração e compartilhamento.
- **VS Code** é o editor no qual você pode escrever, executar e revisar o projeto.

O ciclo de uma alteração passa pelo diretório de trabalho, pela área de preparação (*stage*), por um *commit* local e, quando desejado, pelo repositório remoto. Um `commit` registra uma versão local; ele não publica automaticamente as mudanças no GitHub.

## 2. Ciclo básico de trabalho

Execute comandos no terminal, um grupo por vez. Antes de registrar alterações, examine o estado do repositório e o diff para confirmar que só está incluindo o que pretende.

```bash
# Em uma pasta nova, inicialize o repositório uma única vez.
git init
# Consulte o estado atual do repositório.
git status

# Revise as alterações e selecione os arquivos que irão para o próximo commit.
git diff
git add arquivo.py
git diff --staged

# Registre uma alteração com uma mensagem que explique sua intenção.
git commit -m "Calcula média das notas"

# Consulte o histórico resumido.
git log --oneline
```

Para compartilhar o histórico, conecte o repositório local a um remoto e envie os commits com `git push`. Para trazer mudanças publicadas, use `git pull` conforme o fluxo combinado para o projeto. Em um repositório compartilhado, confira `git status` e a branch atual antes de integrar ou publicar alterações. Não coloque senhas, tokens ou outras credenciais em arquivos versionados.

## 3. Branches e rotina de revisão

Uma branch permite desenvolver uma alteração separadamente da linha principal. Dê à branch um nome curto que indique a tarefa e agrupe cada commit em torno de uma mudança coerente.

Uma rotina simples é: verificar `status` e branch; atualizar conforme a orientação do projeto; programar e testar; revisar `git diff`; preparar os arquivos intencionais; fazer commit; enviar com `push` quando for o momento. Leia a mensagem de erro e o estado do repositório antes de repetir comandos.
