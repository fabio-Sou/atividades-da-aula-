## 1. Pensar antes de programar

Um algoritmo é uma sequência finita e organizada de passos para resolver um problema. Antes do código, identifique:

- **Entradas:** quais dados serão fornecidos?
- **Regras:** quais operações e condições devem ser aplicadas?
- **Saídas:** que resultado precisa ser apresentado?

Uma rotina útil é entender o enunciado, calcular um exemplo à mão, dividir a solução em passos, implementar uma parte de cada vez e testar casos comuns, limites e entradas inesperadas.

```python
# Exemplo: calcular total e média de três medidas.
medida1 = 4.5
medida2 = 7.0
medida3 = 5.5

total = medida1 + medida2 + medida3
media = total / 3

print(f"Total: {total:.2f}")
print(f"Média: {media:.2f}")
```

## 2. Valores, variáveis e tipos

Uma variável associa um nome a um valor. Prefira nomes que indiquem a finalidade do dado (`quantidade`, `media`, `nome_aluno`). Os tipos básicos mais usados no início são:

- `int`: números inteiros, como `12`;
- `float`: números com parte decimal, como `3.5`;
- `str`: texto, como `"Python"`;
- `bool`: valores lógicos `True` ou `False`.

O operador `=` atribui um valor. Para comparar valores, use `==`; `is` é usado principalmente para verificar identidade, por exemplo `valor is None`.

```python
idade = 20             # int
altura = 1.72          # float
nome = "Ana"           # str
matriculada = True     # bool

print(type(idade).__name__, type(altura).__name__)
print(type(nome).__name__, type(matriculada).__name__)
print(idade == 20)
print(None is None)
```

## 3. Expressões e apresentação

Operadores aritméticos comuns: `+`, `-`, `*`, `/`, `//` (divisão inteira), `%` (resto) e `**` (potência). Parênteses ajudam a deixar a ordem das operações explícita. Uma f-string permite inserir valores no texto; por exemplo, `:.2f` mostra duas casas decimais.

`print(..., end="")` troca a quebra de linha final por outro texto. Em geral, uma única f-string é mais simples para montar uma linha.

```python
numerador = 5 + 8 - 4
media = numerador / 3

print(f"{5} + {8} - {4} = {numerador}")
print(f"Média: {media:.2f}")
print(f"Divisão inteira: {17 // 5}; resto: {17 % 5}")
```

## 4. Entrada e conversão

`input()` sempre devolve texto (`str`). Converta para `int` ou `float` quando precisar fazer contas. Texto não vazio convertido diretamente com `bool(...)` resulta em `True`, mesmo se o texto digitado for `"False"`; para uma pergunta sim/não, compare o texto normalizado com respostas aceitas.

```python
nome = input("Digite seu nome: ").strip()
idade = int(input("Digite sua idade: "))
altura = float(input("Digite sua altura em metros: "))

print(f"Olá, {nome}! Você tem {idade} anos e mede {altura:.2f} m.")
```

```python
resposta = input("Você está matriculado? (s/n): ").strip().lower()
matriculado = resposta in ("s", "sim")

print(f"Matriculado: {matriculado}")
```

## 5. Funções úteis e testes

- `len(valor)` conta elementos; em texto, conta caracteres (incluindo espaços).
- `abs(numero)` retorna o valor absoluto.
- `round(numero, casas)` arredonda um número; f-strings controlam como ele é exibido.
- `sorted(iteravel)` cria uma nova lista ordenada.
- `sum(iteravel)` soma valores numéricos.
- `sqrt` pode ser importada de `math` para calcular raiz quadrada.

Teste manualmente resultados esperados. Se uma conversão de entrada puder falhar, `try/except ValueError` permite tratar o dado inválido com uma mensagem clara.

```python
from math import sqrt

valores = [8, 9, 1]
print(f"Quantidade: {len(valores)}")
print(f"Média: {sum(valores) / len(valores):.2f}")
print(f"Raiz de 9: {sqrt(9)}")
print(f"Valor absoluto de -5.3: {abs(-5.3)}")
```
