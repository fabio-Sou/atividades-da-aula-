## 1. Decisões e condições

`if`, `elif` e `else` escolhem caminhos. Uma condição pode combinar comparações com `and` (todas verdadeiras), `or` (ao menos uma verdadeira) e `not` (negação). Para comparar valores, use `==` e `!=`; `in` testa pertinência em uma coleção. `is` não é substituto de `==`: reserve-o sobretudo para `is None`.

Dê atenção aos limites. Se as faixas forem mutuamente exclusivas, estabeleça claramente se cada extremidade está incluída (`<` ou `<=`) e teste exatamente os valores de fronteira.

```python
temperatura = 31

if temperatura > 30:
    print("Leve água")
elif temperatura >= 18:
    print("Temperatura amena")
else:
    print("Está frio")
```

## 2. Repetições

- Use `for` para percorrer itens ou uma sequência conhecida. `range(inicio, fim, passo)` inclui o início e não inclui o fim.
- Use `while` quando a repetição depende de uma condição que pode mudar. Atualize essa condição para evitar um laço infinito.
- `break` encerra o laço; `continue` avança para a próxima repetição; `pass` não faz nada e pode servir como espaço reservado temporário.
- Um **acumulador** guarda um resultado parcial, como uma soma; uma variável de contagem guarda quantas vezes algo ocorreu.

Antes de executar um laço, simule algumas iterações e acompanhe o valor das variáveis que mudam.

```python
# Soma dos inteiros de 1 até 5 usando um acumulador.
soma = 0
for numero in range(1, 6):
    soma += numero

print(soma)
```

## 3. Funções: dividir, reutilizar e testar

`def` define uma função. Os nomes entre parênteses são **parâmetros**; os valores usados ao chamar a função são **argumentos**. `return` devolve um resultado. Se uma função só imprime, ela apresenta informação, mas não devolve automaticamente esse valor para o restante do programa.

Funções pequenas, com nomes claros e uma responsabilidade, ajudam a testar e explicar a solução. Argumentos podem ser posicionais ou nomeados; valores padrão tornam alguns argumentos opcionais. Variáveis criadas dentro da função normalmente têm escopo local. A recursão é uma função que chama a si mesma e precisa de um caso-base para terminar.

```python
def media(valores):
    """Calcula a média de uma coleção não vazia de números."""
    return sum(valores) / len(valores)


notas = [8.0, 9.5, 7.0]
resultado = media(notas)
print(f"Média: {resultado:.2f}")
```

Uma função pode devolver mais de um resultado. Python os agrupa em uma tupla, que pode ser desempacotada em variáveis. Ao dividir, por exemplo, convém verificar que o divisor não seja zero.

```python
def dividir_com_resto(dividendo, divisor):
    if divisor == 0:
        raise ValueError("O divisor não pode ser zero.")
    return dividendo // divisor, dividendo % divisor


quociente, resto = dividir_com_resto(14, 4)
print(quociente, resto)
```

## 4. Coleções: escolha pela necessidade

| Coleção | Use quando... | Exemplo |
| --- | --- | --- |
| `list` | precisa manter ordem e pode alterar/adicionar itens | `notas = [8.0, 9.5]` |
| `tuple` | valores ordenados formam um registro que não deve ser alterado | `ponto = (10, 20)` |
| `set` | precisa de elementos únicos ou comparar grupos | `tags = {"python", "dados"}` |
| `dict` | cada valor deve ser encontrado por uma chave significativa | `aluno = {"nome": "Lia"}` |

Listas usam índices começando em zero. Dicionários acessam valores por chaves. Conjuntos não são uma sequência para indexar. Tuplas são imutáveis, embora possam conter objetos mutáveis.

```python
notas = [8.0, 9.5, 7.0]
notas.append(10.0)
print(f"Primeira nota: {notas[0]}; quantidade: {len(notas)}")

ponto = (10, 20)
x, y = ponto

inscritos_a = {"Ana", "Bia", "Caio"}
inscritos_b = {"Bia", "Davi"}
print(f"Em ambos: {inscritos_a & inscritos_b}")

aluno = {"nome": "Lia", "nota": 9.5}
print(aluno.get("email", "Sem e-mail cadastrado"))
```

## 5. Percorrer, transformar e tratar erros

Percorra diretamente uma lista quando só precisa dos valores. Use `enumerate()` quando também precisar do índice e `dicionario.items()` quando precisar de chave e valor. Fatiamento seleciona partes de uma sequência (`itens[inicio:fim:passo]`); compreensões criam listas de forma concisa, mas devem continuar legíveis.

Erros comuns incluem `IndexError` (índice inexistente), `KeyError` (chave ausente), `TypeError` (operação incompatível) e `ValueError` (valor inválido). Trate apenas exceções esperadas; por exemplo, converta entrada numérica dentro de `try` e capture `ValueError` para explicar o formato esperado.

```python
frutas = ["uva", "maçã", "pera"]
for indice, fruta in enumerate(frutas, start=1):
    print(indice, fruta)

quadrados = [numero ** 2 for numero in range(6)]
print(quadrados)

aluno = {"nome": "Lia", "nota": 9.5}
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
```
