# Guia de estudo — Computação para Engenharia

Este caderno organiza os resumos dos PDFs da pasta de materiais da aula e indica onde encontrar as anotações pessoais antigas. Os resumos foram escritos para revisão: priorizam ideias, exemplos curtos e cuidados comuns, sem substituir a leitura das fontes.

## Ordem sugerida

1. [Fundamentos de Python](./resumos/01_fundamentos_python.ipynb) — algoritmos, tipos, entrada e saída, expressões e testes.
2. [Fluxo, funções e coleções](./resumos/02_fluxo_funcoes_colecoes.ipynb) — decisões, laços, funções e estruturas de dados.
3. [Git, GitHub e Copilot](./resumos/03_git_github_copilot.ipynb) — histórico de versões e uso responsável do assistente.
4. Consulte as anotações em `anotacoes/` para rever os exemplos originais, agora agrupados e acompanhados de esclarecimentos.

## Materiais resumidos

- `Introducao CPE.pdf`, `Fluxo_v1.pdf`, `Funcoes.pdf` e `Colecoes.pdf`: aulas introdutórias de programação e Python.
- `EBOOK - INTRODUÇÃO A PYTHON (EDITORA IFRN).pdf`: referência complementar mais extensa, incluindo aplicações de sistemas operacionais.
- `Plano de Ensino - CPE - Daniel Alves da Silva - 2026.2 (final).pdf`: objetivos, programa e avaliação da disciplina.
- `01_Git_e_GitHub_para_Iniciantes.pdf` e `Engenharia_de_Prompt_com_Copilot_para_Aprender_Python.pdf`: versionamento e apoio ao estudo com IA.

## Como revisar

Leia a explicação, execute cada célula de código e depois tente modificar os exemplos sem consultar a solução. Para cada conceito, teste um caso comum, um caso-limite e uma entrada inesperada. Os notebooks dos módulos e as atividades da disciplina continuam separados.
