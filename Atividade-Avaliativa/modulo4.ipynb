```python
# Questão 2

def box_of_hashes(line):
    for _ in range(line):
        print("#" * 10)

box_of_hashes(5)
print()
box_of_hashes(2)
```

```python
# Questão 3

def square_of_hashes(line):
    for _ in range(line):
        print("#" * line)

square_of_hashes(5)
print()
square_of_hashes(3)
```

```python
# Questão 4

def square(line, x):
    for _ in range(line):
        print(x * line)

square(5, "*")
print()
square(3, "o")
```

```python
# Questão 5

def triangle(line):
    for i in range(1, line + 1):
        print("#" * i)

triangle(6)
print()
triangle(3)
```

```python
# Questão 6

def shape(line, x, y, z):
    for i in range(1, line + 1):
        print(x * i)
    for y in range(1, y + 1):
        print(z * line)

shape(5, "X", 3, "*")
print()
shape(2, "o", 4, "+")
print()
shape(3, ".", 0, ",")
```

```python
# Questão 7

def spruce(x):
    print("a spruce!")
    for i in range(1, x + 1):
        print(" " * (x - i) + "*" * (2 * i - 1))
    print(" " * (x - 1) + "*")

spruce(5)
```

```python
# Questão 8

def greatest_number(x, y, z):
    if x > y and x > z:
        return x
    elif y > x and y > z:
        return y
    else:
        return z 

print(greatest_number(3, 4, 1)) # 4
print(greatest_number(99, -4, 7)) # 99
print(greatest_number(0, 0, 0)) # 0
```

```python
# Questão 9

def same_chars(string1, x, y):
    if y >= len(string1) or x >= len(string1):
        return False
    elif string1[x] == string1[y]:
        return True
    else:
        return False

# mesmos caracteres a e a
print(same_chars("programador", 5, 7)) # True
 
# caracteres diferentes p e r
print(same_chars("programador", 0, 4)) # False
 
# o segundo índice está fora da string
print(same_chars("programador", 0, 12)) # False
```
