**Para revisar:** `strip()` devolve uma string sem espaços nas extremidades, mas não altera a variável original. `len()` conta todos os caracteres, inclusive espaços. Para contar o nome sem espaços extras nas pontas, faça `nome_limpo = nome.strip()` e use `len(nome_limpo)`. Índices começam em zero e o limite final de um fatiamento não é incluído.
