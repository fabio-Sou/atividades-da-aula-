```python
# Questão 31

# Vamos tomar a raiz quadrada do módulo de matemática em uso
from math import sqrt
A = int(input("Valor de a: "))
B = int(input("Valor de b: "))
C = int(input("Valor de c: "))

M = ((B ** 2) + ((- 4 * A )* C))
R1 = (- B + sqrt(M))/ (2 * A)
R2 = (- B - sqrt(M))/ (2 * A)

print(f"As raízes são {R1} e {R2}")


# Note que a raiz quadrada também pode ser calculada usando potência.
# sqrt(9) é equivalente a 9 ** 0.5
```
