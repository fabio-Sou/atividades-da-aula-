```python
# Questão 20

pal = input("Por favor, digite uma palavra: ")
let = input("Por favor, digite um caractere: ")

# Encontra a primeira ocorrência do caractere na palavra.
indice = pal.find(let)
# Confere se encontrou o caractere e se há três caracteres a partir dele.
if indice != -1 and indice + 3 <= len(pal):
    # Imprime os três caracteres a partir da ocorrência encontrada.
    print(pal[indice:indice + 3])
```

```python
# Questão 22

pal = input("Por favor, digite uma string: ")
substring = input("Por favor, digite uma substring: ")

if substring == "":
    print("A substring não pode ser vazia.")
else:
    primeiro = pal.find(substring)
    if primeiro == -1:
        print("A substring não foi encontrada.")
    else:
        # A busca começa depois da primeira substring para evitar sobreposição.
        segundo = pal.find(substring, primeiro + len(substring))
        if segundo == -1:
            print("A substring não ocorre duas vezes na string.")
        else:
            print(f"A segunda ocorrência da substring está no índice {segundo}.")
```

```python
# Questão 33
def chessboard(size):
    for linha in range(size):
        linha_tabuleiro = ""
        for coluna in range(size):
            linha_tabuleiro += "1" if (linha + coluna) % 2 == 0 else "0"
        print(linha_tabuleiro)


#chessboard(3)
#print()
#chessboard(6)
```

```python
# Questão 34
def squared(text, size):
    if not text or size <= 0:
        return

    for linha in range(size):
        inicio = linha * size
        linha_quadrado = "".join(
            text[(inicio + coluna) % len(text)] for coluna in range(size)
        )
        print(linha_quadrado)


#squared("ab", 3)
#print()
#squared("aybabtu", 5)
```
