```python
# ------------------ Exercícios 1 | Fundamento -----------------

# A)Subtração
# Leia dois valores e mostre a diferença do primeiro para o segundo.

valor1 = float(input("Digite o primeiro valor: "))
valor2 = float(input("Digite o segundo valor: "))
diferenca = valor1 - valor2

print(f"A diferença do primeiro para o segundo é: {diferenca}")
```

```python
# B) Produto
# Leia dois valores e apresente o resultado da multiplicação.

valor1 = int(input("Digite o primeiro valor: "))
valor2 = int(input("Digite o segundo valor: "))
produto = valor1 * valor2
print(f"O produto de {valor1} e {valor2} é {produto}.")
```

```python
# C) Hipotenusa
# Leia os dois catetos e calcule a hipotenusa.

import math

cateto1 = float(input("Digite o primeiro cateto: "))
cateto2 = float(input("Digite o segundo cateto: "))
hipotenusa = math.hypot(cateto1, cateto2)
print(f"A hipotenusa de {cateto1} e {cateto2} é {hipotenusa}.")
```

```python
# D) Total e média
# Leia três medidas; mostre soma e média com duas casas.

medida1 = float(input("Digite a primeira medida: "))
medida2 = float(input("Digite a segunda medida: "))
medida3 = float(input("Digite a terceira medida: "))
soma = medida1 + medida2 + medida3
media = soma / 3
print(f"A soma das medidas é {soma} e a média é {media:.2f}.")
```

```python
# ------------------- Exercícios 2 | Decisões -----------------

# A Maior e intermediário
# Leia três valores distintos e informe o maior e o valor do meio.

maior = 0
valor1 = int(input("Digite o primeiro valor: "))
valor2 = int(input("Digite o segundo valor: "))
valor3 = int(input("Digite o terceiro valor: "))
print(f"Os valores digitados foram: {valor1}, {valor2} e {valor3}.")

if valor1 > valor2 and valor1 > valor3:
    maior = valor1
    if valor2 > valor3:
        intermediario = valor2
    else:
        intermediario = valor3
elif valor2 > valor1 and valor2 > valor3:
    maior = valor2
    if valor1 > valor3:
        intermediario = valor1
    else:
        intermediario = valor3
else:
    maior = valor3
    if valor1 > valor2:
        intermediario = valor1
    else:
        intermediario = valor2
print(f"O maior valor é {maior} e o valor intermediário é {intermediario}.")
```

```python
#B) Classificador
#Informe se um inteiro é positivo, negativo ou zero; se não for zero, diga se é par.

inteiro = int(input("Digite um número inteiro: "))
if inteiro > 0:
    print(f"O número {inteiro} é positivo.")
    if inteiro % 2 == 0:
        print(f"O número {inteiro} é par.")
    else:
        print(f"O número {inteiro} é ímpar.")
elif inteiro < 0:
    print(f"O número {inteiro} é negativo.")
else:
    print(f"O número {inteiro} é zero.")
```

```python
# C) Triângulo
# Valide os lados pela desigualdade triangular e, se válido, calcule a área de Heron.

lado1 = float(input("Digite o comprimento do primeiro lado: "))
lado2 = float(input("Digite o comprimento do segundo lado: "))
lado3 = float(input("Digite o comprimento do terceiro lado: "))

if (lado1 + lado2 > lado3) and (lado1 + lado3 > lado2) and (lado2 + lado3 > lado1):
    print("Os lados formam um triângulo válido.")
    s = (lado1 + lado2 + lado3) / 2
    area = math.sqrt(s * (s - lado1) * (s - lado2) * (s - lado3))
    print(f"A área do triângulo é {area:.2f}.")
```

```python
# D) Tarifa revisada
# Adapte o aluguel: taxa mínima de R$ 30 e 5% de desconto acima de 300 km.

distancia = float(input("Digite a distância percorrida em km: "))
valor_por_km = float(input("Digite o valor cobrado por km: R$ "))

tarifa = distancia * valor_por_km

if tarifa < 30:
    tarifa = 30

if distancia > 300:
    tarifa = tarifa * 0.95

print(f"A tarifa final é R$ {tarifa:.2f}.")
```

```python
# ------------------- Exercícios 3 | Strings -----------------

# A) Vogais
# Leia uma palavra de cinco caracteres; conte e liste as vogais encontradas.

palavra = input("Digite uma palavra com cinco caracteres: ").strip()

if len(palavra) != 5:
    print("A palavra deve ter exatamente cinco caracteres.")
else:
    vogais = [letra for letra in palavra.lower() if letra in "aeiou"]
    print(f"Vogais encontradas: {vogais}")
    print(f"Quantidade de vogais: {len(vogais)}")
```

```python
# B) Cifra simples
# Troque a -> i, e -> o e i -> u; preserve os demais caracteres.

texto = input("Digite um texto: ")
traducao = str.maketrans({"a": "i", "e": "o", "i": "u"})
texto_cifrado = texto.translate(traducao)

print(f"Texto cifrado: {texto_cifrado}")
```

```python
# C) Palíndromo
# Ignore espaços e maiúsculas; verifique se o texto é igual ao inverso.

texto = input("Digite um texto: ")
texto_normalizado = "".join(texto.lower().split())

if texto_normalizado == texto_normalizado[::-1]:
    print("O texto é um palíndromo.")
else:
    print("O texto não é um palíndromo.")
```

```python
# D) Subnúmero
# Leia p e q como texto; informe se a sequência p aparece dentro de q.

p = input("Digite a sequência p: ")
q = input("Digite a sequência q: ")

if p in q:
    print(f"A sequência {p!r} aparece dentro de {q!r}.")
else:
    print(f"A sequência {p!r} não aparece dentro de {q!r}.")
```

```python
# ------------------ Desafios de engenharia ------------------
# A) Decibéis
# Com P1 = 1 W e P2 > 0, calcule 10 * log10(P2 / P1).

import math

p1 = 1
p2 = float(input("Digite P2 em watts: "))

if p2 > 0:
    decibeis = 10 * math.log10(p2 / p1)
    print(f"A intensidade sonora é {decibeis:.2f} dB.")
else:
    print("P2 deve ser maior que zero.")
```

```python
# B) Área de Heron
# Receba três lados, valide o triângulo e calcule a área.

lado1 = float(input("Digite o primeiro lado: "))
lado2 = float(input("Digite o segundo lado: "))
lado3 = float(input("Digite o terceiro lado: "))

if (lado1 > 0 and lado2 > 0 and lado3 > 0
        and lado1 + lado2 > lado3
        and lado1 + lado3 > lado2
        and lado2 + lado3 > lado1):
    semiperimetro = (lado1 + lado2 + lado3) / 2
    area = math.sqrt(
        semiperimetro
        * (semiperimetro - lado1)
        * (semiperimetro - lado2)
        * (semiperimetro - lado3)
    )
    print(f"A área do triângulo é {area:.2f}.")
else:
    print("Os lados não formam um triângulo válido.")
```

```python
# C) Quatro dígitos
# Encontre números cuja raiz seja igual à soma das duas metades.

numeros_encontrados = []

for numero in range(1000, 10000):
    raiz = math.isqrt(numero)

    if raiz * raiz == numero:
        texto = str(numero)
        primeira_metade = int(texto[:2])
        segunda_metade = int(texto[2:])

        if primeira_metade + segunda_metade == raiz:
            numeros_encontrados.append((numero, raiz))

for numero, raiz in numeros_encontrados:
    print(f"{numero}: raiz = {raiz}, soma das metades = {raiz}")
```

```python
# D) Número primo
# Informe se n > 1 tem algum divisor de 2 até √n.

n = int(input("Digite um número inteiro: "))
tem_divisor = False

if n > 1:
    for divisor in range(2, math.isqrt(n) + 1):
        if n % divisor == 0:
            tem_divisor = True
            break

if n > 1 and not tem_divisor:
    print(f"{n} é primo.")
else:
    print(f"{n} não é primo.")
```
