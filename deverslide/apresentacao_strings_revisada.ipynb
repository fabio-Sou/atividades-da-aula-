```python
# Questão 1 — Contando caracteres
# Leia e responda
# • Qual número será exibido?
# • Explique se o espaço entre as palavras entra na contagem

texto = "Bom dia"
print(len(texto))

# o espaço entre ele vaiser considerado já que não tem nada que de ser contado, como por um .strip() que tira os espaços em branco do início e do 
# final da string, mas para tirar os espaços do meio da string, teria que ser feito um .replace(" ", "") para substituir os espaços por nada, e assim 
# não contar os espaços entre as palavras.

print(texto.replace(" ", "")) # substitui os espaços por nada, e assim não conta os espaços entre as palavras
print(len(texto.replace(" ", "")))
```

```python
# Questão 2 — Limpando e padronizando
# Observe a sequência de operações
# • Qual texto será exibido entre os colchetes?
# • Explique o que strip() e upper() fazem.

nome = " Ana Silva "
resultado = nome.strip().upper() # strip() tira os espaços em branco do início e do final da string, e upper() transforma a string em maiúsculo
print("[" + resultado + "]")
```

```python
# Questão 3 — Buscando e contando
# Descubra as três saídas
# • Escreva o resultado de cada print(), na ordem.
# • Lembre-se: os índices começam em zero.

fruta = "banana"
print(fruta.find("na")) # 2       # .find() retorna o índice da primeira ocorrência da substring "na" na string "banana", que é 2.
print(fruta.count("a")) # 3       # .count() retorna o número de ocorrências da substring "a" na string "banana", que é 3.
print("maçã" in fruta)  # false   # O operador "in" verifica se a substring "maçã" está presente na string "banana". Como não está, o resultado é False.
```

```python
# Questão 4 — Separando e juntando
# Complete os dois espaços
# • Separe as cores usando a vírgula.
# • Depois, junte as cores usando um hífen.

cores = "azul,verde,rosa"

# lista = cores.________(",")
lista = cores.split(",")           # .split() separa a string em uma lista de substrings, usando a vírgula como delimitador
print(lista)

# resultado = "-".________(lista)
resultado = "-".join(lista)        # .join() junta os elementos da lista em uma única string, usando o hífen como delimitador
print(resultado)
```

```python
# Questão 5 — Pequeno cadastro
# Escreva um programa em Python
# • Peça um nome de usuário com input().
# • Remova os espaços do início e do fim.
# • Transforme o nome em letras minúsculas.
# • Exiba o nome tratado e sua quantidade de caracteres.

# Entrada de exemplo: " ALUNO01 "
# Saída esperada:
# Usuário: aluno01
# Caracteres: 7

nomr = input("Digite seu nome: ")
nome_tratado = nomr.strip().lower()        # .lower() transforma a string em minúsculo
print(f"Usuário: {nome_tratado}")
print(f"Caracteres: {len(nome_tratado)}")
```
